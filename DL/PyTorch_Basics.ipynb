# PyTorch Basics Workbook
Learn tensors and basic concepts.

```python
import torch
```

## Create Tensor

```python
x=torch.tensor([1,2,3])
print(x)
```

## Data Type

```python
print(x.dtype)
```

## Dimensions

```python
a=torch.tensor(5)
b=torch.tensor([1,2])
c=torch.tensor([[1,2],[3,4]])
print(a.ndim,b.ndim,c.ndim)
```

## Shape

```python
print(c.shape)
```

## Zeros Ones Random

```python
print(torch.zeros(2,3))
print(torch.ones(2,3))
print(torch.rand(2,3))
```

## Arithmetic

```python
a=torch.tensor([1,2,3]); b=torch.tensor([4,5,6])
print(a+b,a-b,a*b,a/b)
```

## Matrix Multiplication

```python
A=torch.tensor([[1,2],[3,4]])
B=torch.tensor([[5,6],[7,8]])
print(torch.matmul(A,B))
```

## Reshape

```python
x=torch.arange(12)
print(x.reshape(3,4))
```

## Indexing

```python
x=torch.tensor([[10,20,30],[40,50,60]])
print(x[0],x[:,1])
```

## Statistics

```python
x=torch.tensor([10.,20.,30.,40.])
print(torch.mean(x),torch.sum(x),torch.max(x))
```

## Autograd

```python
x=torch.tensor(2.0,requires_grad=True)
y=x*x
y.backward()
print(x.grad)
```

## Simple NN

```python
import torch.nn as nn
model=nn.Sequential(nn.Linear(3,5),nn.ReLU(),nn.Linear(5,1))
print(model)
```
