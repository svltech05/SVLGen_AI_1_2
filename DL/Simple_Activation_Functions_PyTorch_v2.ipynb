# Activation Functions using PyTorch
Includes Softmax.

```python
import torch
import torch.nn as nn
import matplotlib.pyplot as plt

x=torch.linspace(-5,5,200)
```

## Sigmoid

Purpose, advantages, limitations and use cases can be discussed alongside this plot.

```python
act=nn.Sigmoid()
y=act(x)
plt.figure(figsize=(6,4))
plt.plot(x.numpy(),y.numpy(),linewidth=2)
plt.title("Sigmoid")
plt.grid(True)
plt.show()
print(act(torch.tensor([-3.,-1.,0.,1.,3.])))
```

## Tanh

Purpose, advantages, limitations and use cases can be discussed alongside this plot.

```python
act=nn.Tanh()
y=act(x)
plt.figure(figsize=(6,4))
plt.plot(x.numpy(),y.numpy(),linewidth=2)
plt.title("Tanh")
plt.grid(True)
plt.show()
print(act(torch.tensor([-3.,-1.,0.,1.,3.])))
```

## ReLU

Purpose, advantages, limitations and use cases can be discussed alongside this plot.

```python
act=nn.ReLU()
y=act(x)
plt.figure(figsize=(6,4))
plt.plot(x.numpy(),y.numpy(),linewidth=2)
plt.title("ReLU")
plt.grid(True)
plt.show()
print(act(torch.tensor([-3.,-1.,0.,1.,3.])))
```

## Leaky ReLU

Purpose, advantages, limitations and use cases can be discussed alongside this plot.

```python
act=nn.LeakyReLU(0.1)
y=act(x)
plt.figure(figsize=(6,4))
plt.plot(x.numpy(),y.numpy(),linewidth=2)
plt.title("Leaky ReLU")
plt.grid(True)
plt.show()
print(act(torch.tensor([-3.,-1.,0.,1.,3.])))
```

## ELU

Purpose, advantages, limitations and use cases can be discussed alongside this plot.

```python
act=nn.ELU()
y=act(x)
plt.figure(figsize=(6,4))
plt.plot(x.numpy(),y.numpy(),linewidth=2)
plt.title("ELU")
plt.grid(True)
plt.show()
print(act(torch.tensor([-3.,-1.,0.,1.,3.])))
```

# Softmax

Softmax is different from the other activation functions.

- **Purpose:** Converts a vector of scores (logits) into probabilities.
- **Output Range:** 0 to 1
- **Property:** The probabilities always sum to **1**.
- **Use Case:** Output layer of **multi-class classification** problems.
- **Examples:** Digit recognition (0-9), animal classification, flower classification.


```python
softmax = nn.Softmax(dim=0)

scores = torch.tensor([2.5, 4.0, 1.5])

print("Raw Scores :", scores)

prob = softmax(scores)

print("Softmax Probabilities :", prob)
print("Sum of Probabilities :", prob.sum())
```
