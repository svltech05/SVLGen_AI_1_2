```python
import numpy as np
f=open('D:/Ashwath/NS_Trainings/MyStuff/diabetic.txt')
h = f.readline()
lines = f.readlines()
age = []
wgt = []
hgt = []
y = []
for line in lines:
    w=line.strip().lower().split(',')
    age.append(float(w[1]))
    wgt.append(float(w[2]))
    hgt.append(float(w[3]))
    y.append(int(w[-1]))
X = np.c_[age,wgt,hgt]
print(X)    
print(y)
```

```python
from sklearn.svm import SVC
model = SVC()
model.fit(X,y)
ycap=model.predict(X)
print(ycap)
print(y)
```

```python
from sklearn.metrics import accuracy_score
acc=accuracy_score(y,ycap)
print("Accuracy of SVM : ",acc*100)
```

```python
#comparing accuracy with different models
# navie bayes

from sklearn.naive_bayes import GaussianNB
model = GaussianNB()
model.fit(X,y)
nbycap=model.predict(X)
nbacc=accuracy_score(y,nbycap)

print("Accuracy of Navie Bayes : ",nbacc*100)
```

```python
#decision tree

from sklearn.tree import DecisionTreeClassifier
model=DecisionTreeClassifier()
model.fit(X,y)
dtycap=model.predict(X)
dtacc=accuracy_score(y,dtycap)
print("Accuracy of Decision Tree : ",dtacc*100)
```

```python
# logistic regression

from sklearn.linear_model import LogisticRegression
model = LogisticRegression()
model.fit(X,y)
lgycap=model.predict(X)
lgacc = accuracy_score(y,lgycap)
print("Accuracy of Logistic Regression : ",lgacc*100)
print(lgycap)
print(y)
```
