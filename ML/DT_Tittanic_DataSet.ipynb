```python
#Before transforming data into numeric ones, we might want to do some 
#feature engineering, i.e. features like Name, Ticket, Cabin and Embarked 
#do not have any impact on the survival status of the passengers. 
#Name, Ticket, Cabin and Embarked can be dropped 

train = train.drop(['Name','Ticket', 'Cabin','Embarked'], axis=1)

test = test.drop(['Name','Ticket', 'Cabin','Embarked'], axis=1)
```

```python
print(train.head())
```

```python
#Enconding
labelEncoder = LabelEncoder()
labelEncoder.fit(train['Sex'])
labelEncoder.fit(test['Sex'])
train['Sex'] = labelEncoder.transform(train['Sex'])
test['Sex'] = labelEncoder.transform(test['Sex'])
```

```python
#print(train['Sex'])
#train.info()
train.head()
#test.head()
```

```python
#creating input matrix and label matrix
X = np.array(train.drop(['Survived'], axis=1).astype(float))
#X = np.array(train.drop(['Survived'],1).astype(float))
print(X)
```

```python
Xtest =  np.array(test.astype(float))
print(Xtest)
```

```python
Y = np.array(train['Survived'])
print(Y)
```

```python
#callaing library
from sklearn.tree import DecisionTreeClassifier

model=DecisionTreeClassifier()

model.fit(X,Y)

ycap=model.predict(X)

#print(ycap)
```

```python
from sklearn.metrics import accuracy_score
acc = accuracy_score(Y,ycap)
print (acc*100)
```
