Input Data: Each customer is represented as a point with features (Income, Spending Score).

Choose k: Suppose we pick k = 3

clusters.

Clustering:

Cluster 1: High income, high spending → “Premium Shoppers”

Cluster 2: Moderate income, moderate spending → “Average Shoppers”

Cluster 3: Low income, low spending → “Budget Shoppers”

Centroids: Each cluster has a “center” (average income & spending score of that group).

Visualization: When plotted, customers form groups with different colors, and centroids are marked with red X’s.

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
```

```python
# Example dataset
data = pd.DataFrame({
    'Age': [25, 45, 35, 23, 52, 40, 60, 30],
    'Annual_Income': [40000, 80000, 60000, 35000, 90000, 70000, 30000, 50000],
    'Spending_Score': [80, 40, 60, 90, 30, 50, 20, 70]
})
```

```python
# Select features for clustering
X = data[['Annual_Income', 'Spending_Score']]
```

```python
# Scatter plot BEFORE clustering 
plt.scatter(X['Annual_Income'], X['Spending_Score'], c='blue', marker='o') 
plt.xlabel("Annual Income") 
plt.ylabel("Spending Score") 
plt.title("Customer Data Before Clustering") 
plt.grid(True) 
plt.show()
```

```python
# Apply K-Means with 3 clusters
kmeans = KMeans(n_clusters=3, random_state=0)
data['Cluster'] = kmeans.fit_predict(X)
```

```python
# Visualize clusters
plt.scatter(X['Annual_Income'], X['Spending_Score'], c=data['Cluster'], cmap='viridis')
plt.xlabel("Annual Income")
plt.ylabel("Spending Score")
plt.title("Customer Segmentation with K-Means")
plt.show()
```

```python
labels = kmeans.labels_ 
centroids = kmeans.cluster_centers_
print(labels)
print(centroids)
```
