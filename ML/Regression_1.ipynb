```python
import numpy as np

def predict(x,b):
    return x.dot(b)

def mse(y,ycap):
    return((y-ycap)**2).mean()

def derivative(x,y,beta):
    ycap = predict(x,beta)
    return(x.T.dot(y-ycap))/y.size

#standard deviation
def sd(x):     
    return (((x-x.mean())**2).sum() / (x.size-1))**0.5 

#standard scaling
def scale(x):  
    return (x-x.mean())/sd(x)
   
def train(x,y,iter,alpha,conv = 0.0000001):
    beta = np.c_[np.zeros(x.shape[1])]
    perr = 0
    j = 0
    for i in range(iter):
        ycap = predict(x,beta)
        cerr = mse(y,ycap)
        if i % 50 == 0:
            print("Current Error ",cerr)
        diff = np.abs(perr-cerr)
        if diff <= conv:
            print("Training Completed ",i+1, "iterations")
            j = 1
            break
        delta = derivative(x,y,beta)
        beta = beta + alpha * delta
        perr = cerr
    if j == 0:
        print("Training not completed")
    return beta

def scalematrix(x):
    ncol=x.shape[1]   #1,60,5.5,40  x = (3,5) x[1]  =5
    for i in range(ncol):  
        if i == 0:
            continue
        c = x[:,i]
        x[:,i] = scale(c) 
    return x                 

def accuarcy(y,ycap,closeness):
    de = 100-closeness
    d = np.abs((y-ycap)/y)*100
    pcnt = d[d<=de].size
    n=y.size
    return pcnt/n*100
```

```python
f=open("D:/Ashwath/NS_Trainings/MyStuff/paitents.txt")
header=f.readline()
lines=f.readlines()
#print(lines)
age=[]
wgt=[]
hgt=[]
chol=[]
for line in lines:
    l=line.strip().split(',')
    #print(l)
    a=float(l[2])
    w=float(l[3])
    h=float(l[-2])
    ch=float(l[-1])
    age.append(a)
    wgt.append(w)
    hgt.append(h)
    chol.append(ch)
print(age,wgt,hgt,chol)

ones=np.ones(len(lines))
X=np.c_[ones,age,wgt,hgt] #input matrix
print(X)

#print(chol)
Y=np.c_[chol] #label / output matrix
print(Y)
```

```python
age=np.array(age)
hgt=np.array(hgt)
wgt=np.array(wgt)
chol=np.array(chol)

mage = age.mean()
mwgt = wgt.mean()
mhgt = hgt.mean()
mchol=chol.mean()

sage = sd(age)
swgt = sd(wgt)
shgt = sd(hgt)
schol = sd(chol)
```

```python
print("Before Scaling :")
print(X)
print(Y)
print("="*50)
print("After Scaling :")
X = scalematrix(X)
Ys = scale(Y)
print(X)
print(Ys)
```

```python
theta = train(X,Ys,10000,0.05)
```

```python
print(theta)
```

```python
ycap = predict(X,theta)
print("============================")
print (ycap)
```

```python
#unscaled prediction

# def scale(x):  
#     return (x-x.mean())/sd(x)

uycap = ycap * schol + mchol
print(uycap)
```
