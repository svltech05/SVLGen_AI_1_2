```python
import numpy as np

#inv(X.T.dot(x)).dot(X.T.(dot(Y))
def coeffs(x,y):
    from numpy.linalg import inv
    left = inv(x.T.dot(x))
    right = x.T.dot(y)
    beta = left.dot(right)
    return beta

#Y = 1 / 1 + e-XB
def logistic(x,b):
    pre =  1/(1+np.exp(-x.dot(b)))
    return pre

def acc_clf(y,ycap):    #y = 1,0,0,1,1   ycap = 1,0,1,0,1
    r=y==ycap           # r = [t,t,f,f,t]  
    pcnt = r[r==True].size  #r[r==True].size  = 3
    n=y.size #5
    acc=pcnt/n*100  #3/5 * 100
    return acc
```

```python
f=open('D:/Ashwath/NS_Trainings/MyStuff/diabetic.txt')
hd=f.readline()
lines=f.readlines()
age=[]
wgt=[]
hgt=[]
dstat=[]
for line in lines:
    w=line.strip().split(',')
    age.append(float(w[1]))
    wgt.append(float(w[2]))
    hgt.append(float(w[-2]))
    dstat.append(float(w[-1]))
    ones=np.ones(len(lines))    
X=np.c_[ones,age,wgt,hgt]
Y=np.c_[dstat]  
print(X)    
print(Y)    
```

```python
beta = coeffs(X,Y)
print(beta)
```

```python
ycap=logistic(X,beta)
print(ycap)
```

```python
ycap[ycap>0.6]=1
ycap[ycap<0.6]=0
print(ycap)
```

```python
# logistic regression using sklearn library
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

model = LogisticRegression()
model.fit(X,Y)

lgycap=model.predict(X)

lgacc = accuracy_score(Y,lgycap)
print("Accuracy of Logistic Regression : ",lgacc*100)
print(lgycap)
print(Y)
```
