The Breast Cancer dataset contains features computed from digitized images of fine needle aspirates of breast masses. Each sample describes characteristics of cell nuclei, such as radius, texture, smoothness, etc.

Goal: Use K-Means clustering to group patients into clusters (e.g., benign vs malignant).

```python
from sklearn.datasets import load_breast_cancer
import pandas as pd

# Load dataset
data = load_breast_cancer()
df = pd.DataFrame(data.data, columns=data.feature_names)

# Add target labels (benign=0, malignant=1)
df['target'] = data.target
df.head()
```

```python
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

# Select features
X = df[['mean radius', 'mean texture']]
```

```python
# Normalize
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
# Scatter plot BEFORE clustering 
plt.scatter(X_scaled[:,0], X_scaled[:,1], c='blue', marker='o') 
plt.xlabel("Mean Radius") 
plt.ylabel("Mean Texture") 
plt.title("Breast Cancer Data Before Clustering") 
plt.grid(True) 
plt.show()
```

```python
# Apply K-Means
kmeans = KMeans(n_clusters=2, random_state=0)
labels = kmeans.fit_predict(X_scaled)

print("Cluster labels:", labels)
print("Centroids:", kmeans.cluster_centers_)
```

```python
import matplotlib.pyplot as plt

plt.scatter(X_scaled[:,0], X_scaled[:,1], c=labels, cmap='tab10')
plt.scatter(kmeans.cluster_centers_[:,0], kmeans.cluster_centers_[:,1],
            c='red', marker='X', s=200)
plt.xlabel("Mean Radius (scaled)")
plt.ylabel("Mean Texture (scaled)")
plt.title("K-Means Clustering on Breast Cancer Data")
plt.show()
```
