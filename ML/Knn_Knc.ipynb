```python
#implementation of KNN
import numpy as np
f= open("D:/Ashwath/NS_Trainings/MyStuff/knndata.txt")
header = f.readline()
lines = f.readlines()
info=[]
for line in lines:
    w = line.strip().split(",") 
    ins=[int(v) for v in w[:-1]]
    info.append((np.array(ins),w[-1])) 
for v in info:
    print(v)
```

```python
#finding distance
new = np.array([20,35])
dist =[]
for v in info:
    attr = v[0]
    d = ((new - attr)**2).sum()**0.5 
    dist.append((d,v[1]))
dist.sort()
for d in dist:
    print(d)
```

```python
#finding frequency of classifiers
k = 5   # taking 5 as nearst nieghbours
x = dist[:k]
freq = {}        
for d in x:
    if freq.get(d[1])==None: 
        freq[d[1]]=1       
    else:
        freq[d[1]]+=1
print(freq)
```

```python
#define predict function
def predict(data,new): 
    dist =[]
    for v in data:
        attr = v[0]
        d = ((new - attr)**2).sum()**0.5
        #sqrt[(x1 - x2)^2 + (y1 - y2)^2]  
        dist.append((d,v[1]))
    dist.sort()    
        
    k = 5   # taking 5 as nearst nieghbours
    x = dist[:k]       
        
    freq = {}
    for d in x:
        if freq.get(d[1])==None:
            freq[d[1]]=1
        else:
            freq[d[1]]+=1
    return (freq)
```

```python
newdata = np.array([20,35])
prevals = predict(info,newdata)
print (prevals)
```

```python
#implementation of KNC :
infor = {}
for v in info:
    attr = v[0]
    k = v[1]                  
    if infor.get(k)==None:         
        infor[k] = [attr]        
    else:                           
        infor[k].append(attr)           
print(infor)                       

ins={}
for k in infor:
    ins[k]=np.array(infor[k])
print(ins)
```

```python
centorids = {}
for k in ins:
    v = ins[k]
    centorids[k]=v.mean(axis=0)
print (centorids)
```

```python
new = np.array([20,35])
ndist = {}
for k in centorids:
    d=((centorids[k]-new)**2).sum()**0.5
    ndist[k]=d
print (ndist)
```
