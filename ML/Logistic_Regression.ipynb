# Logistic Regression

## 1. What is Logistic Regression?

Logistic Regression is a **supervised machine learning algorithm** mainly used for **binary classification**.

Examples:

- Email → Spam / Not Spam
- Customer → Will Buy / Will Not Buy
- Student → Pass / Fail
- Patient → Positive / Negative
- Transaction → Fraud / Not Fraud

For binary classification, the target variable normally has two classes:

```text
0 → Negative class
1 → Positive class
```

### Basic idea

Logistic Regression first calculates a linear combination of the input features and then passes that value through the **Sigmoid function**.

The output is a probability between **0 and 1**.


## 2. Linear Regression vs Logistic Regression

### Linear Regression

Linear Regression predicts a continuous numerical value.

Examples:

- House price = ₹50 lakh
- Temperature = 32.5°C
- Salary = ₹60,000

Its output can be any real number.

### Logistic Regression

Logistic Regression predicts the probability of belonging to a class.

Example:

```text
Probability of passing = 0.87
```

Using a threshold of 0.5:

```text
0.87 >= 0.5 → Class 1 → Pass
```

So Logistic Regression converts a probability into a class label.


## 3. The Logistic Regression Equation

First, the model calculates:

$$z = b_0 + b_1x_1 + b_2x_2 + \cdots + b_nx_n$$

Then the value `z` is passed through the **Sigmoid function**:

$$\sigma(z) = \frac{1}{1 + e^{-z}}$$

The final prediction is:

$$P(y=1|x) = \sigma(z)$$

### Why Sigmoid?

The Sigmoid function converts any number into a value between **0 and 1**.

```text
Large negative z → probability close to 0
z = 0             → probability = 0.5
Large positive z → probability close to 1
```


````python
import numpy as np
import matplotlib.pyplot as plt

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

z = np.linspace(-10, 10, 200)
probability = sigmoid(z)

plt.figure(figsize=(9, 5))
plt.plot(z, probability)
plt.axhline(0.5, linestyle="--")
plt.axvline(0, linestyle="--")
plt.xlabel("z")
plt.ylabel("Sigmoid Probability")
plt.title("Sigmoid Function")
plt.grid(True)
plt.show()
````

## 4. Understanding the Sigmoid Function

Notice that:

- When `z = -10`, probability is almost 0
- When `z = 0`, probability is exactly 0.5
- When `z = 10`, probability is almost 1

This is useful because classification needs probabilities that can be interpreted between 0 and 1.

### Classification threshold

A common threshold is **0.5**:

```text
Probability >= 0.5 → Class 1
Probability <  0.5 → Class 0
```

The threshold can be changed depending on the application.


````python
# Let's calculate a few sigmoid values

values = [-5, -2, -1, 0, 1, 2, 5]

for value in values:
    print(f"z = {value:>2}  -> probability = {sigmoid(value):.4f}")
````

## 5. Example: Predicting Whether a Student Passes

Suppose we want to predict whether a student passes based on the number of hours studied.

Our dataset could look like:

| Hours Studied | Result |
|---:|---:|
| 1 | 0 |
| 2 | 0 |
| 3 | 0 |
| 4 | 0 |
| 5 | 1 |
| 6 | 1 |
| 7 | 1 |
| 8 | 1 |

Here:

```text
0 → Fail
1 → Pass
```

The model learns the relationship between **Hours Studied** and **Result**.


````python
import numpy as np

hours = np.array([1, 2, 3, 4, 5, 6, 7, 8]).reshape(-1, 1)
result = np.array([0, 0, 0, 0, 1, 1, 1, 1])

print("Hours:")
print(hours.ravel())

print("\nResult:")
print(result)
````

## 6. Train Logistic Regression with Scikit-learn

We will use `LogisticRegression` from `sklearn.linear_model`.

The general workflow is:

```text
Load data
   ↓
Split data into X and y
   ↓
Create model
   ↓
Train model using fit()
   ↓
Predict using predict()
   ↓
Evaluate the model
```


````python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression()

model.fit(hours, result)

print("Model trained successfully!")
````

````python
# Model parameters

print("Intercept:", model.intercept_)
print("Coefficient:", model.coef_)
````

## 7. Understanding Coefficients

For one feature, the equation is:

$$z = b_0 + b_1x$$

where:

- `b0` = intercept
- `b1` = coefficient
- `x` = input feature

If the coefficient is **positive**, increasing the feature generally increases the probability of class 1.

If the coefficient is **negative**, increasing the feature generally decreases the probability of class 1.

The coefficient itself is expressed in terms of **log-odds**, not directly as a percentage increase in probability.


````python
# Predict class labels

predictions = model.predict(hours)

print("Actual:    ", result)
print("Predicted: ", predictions)
````

````python
# Predict probabilities

probabilities = model.predict_proba(hours)

for h, p, pred in zip(hours.ravel(), probabilities[:, 1], predictions):
    print(f"Hours = {h} -> P(Pass) = {p:.4f} -> Prediction = {pred}")
````

## 8. `predict()` vs `predict_proba()`

This is an important distinction.

### `predict()`

Returns the final class:

```python
model.predict(X)
```

Example:

```text
[0, 0, 1, 1]
```

### `predict_proba()`

Returns probabilities for each class:

```python
model.predict_proba(X)
```

Example:

```text
Class 0 probability | Class 1 probability
        0.20         |        0.80
```

For binary classification, the second column (`[:, 1]`) is usually the probability of class 1.


````python
# Predict a new student

new_student = np.array([[4.5]])

prediction = model.predict(new_student)[0]
probability = model.predict_proba(new_student)[0, 1]

print("Predicted class:", prediction)
print("Probability of passing:", round(probability, 4))
````

## 9. Visualizing the Model

The blue points represent the actual observations.

The curve represents the predicted probability of class 1.


````python
x_plot = np.linspace(1, 8, 200).reshape(-1, 1)
y_prob = model.predict_proba(x_plot)[:, 1]

plt.figure(figsize=(9, 5))
plt.scatter(hours, result, label="Actual Data")
plt.plot(x_plot, y_prob, label="Predicted Probability")
plt.axhline(0.5, linestyle="--", label="Threshold = 0.5")
plt.xlabel("Hours Studied")
plt.ylabel("Probability / Class")
plt.title("Logistic Regression — Student Pass Prediction")
plt.legend()
plt.grid(True)
plt.show()
````

## 10. A More Realistic Dataset

Now we will use the built-in **Breast Cancer Wisconsin dataset** from Scikit-learn.

This is a binary classification problem.

> For teaching purposes, this notebook uses the dataset already bundled with Scikit-learn, so no external dataset download is required.


````python
from sklearn.datasets import load_breast_cancer

data = load_breast_cancer()

X = data.data
y = data.target

print("X shape:", X.shape)
print("y shape:", y.shape)
print("Number of features:", X.shape[1])
print("Classes:", data.target_names)
````

## 11. Train-Test Split

We should not train and evaluate a model on exactly the same data.

Instead:

```text
Training data → used to learn patterns
Testing data  → used to evaluate generalization
```

We commonly use an 80/20 split.

`random_state` makes the split reproducible.


````python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training samples:", X_train.shape[0])
print("Testing samples :", X_test.shape[0])
````

## 12. Feature Scaling

Logistic Regression can benefit from feature scaling, especially when features have very different numerical ranges.

We will use `StandardScaler`.

It transforms a feature approximately so that:

- Mean ≈ 0
- Standard deviation ≈ 1

### Important

Fit the scaler **only on training data**.

Then use that fitted scaler to transform both training and testing data.

This prevents information from the test set leaking into the training process.


````python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("Scaling completed.")
````

## 13. Train the Final Logistic Regression Model


````python
final_model = LogisticRegression(max_iter=1000)

final_model.fit(X_train_scaled, y_train)

print("Final model trained successfully!")
````

## 14. Make Predictions


````python
y_pred = final_model.predict(X_test_scaled)
y_prob = final_model.predict_proba(X_test_scaled)[:, 1]

print("First 10 predicted classes:", y_pred[:10])
print("First 10 probabilities:", np.round(y_prob[:10], 4))
````

## 15. Model Evaluation

Important classification metrics include:

### Accuracy

Percentage of all predictions that are correct.

$$Accuracy = \frac{TP + TN}{TP + TN + FP + FN}$$

### Precision

Of all observations predicted as positive, how many were actually positive?

$$Precision = \frac{TP}{TP + FP}$$

### Recall

Of all actual positive observations, how many did the model correctly identify?

$$Recall = \frac{TP}{TP + FN}$$

### F1-score

Harmonic mean of precision and recall.

$$F1 = 2 \times \frac{Precision \times Recall}{Precision + Recall}$$


````python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)

print(f"Accuracy : {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall   : {recall:.4f}")
print(f"F1-score : {f1:.4f}")
````

````python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

cm = confusion_matrix(y_test, y_pred)

print(cm)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=data.target_names
)

disp.plot()
plt.title("Confusion Matrix")
plt.show()
````

## 17. Classification Report

Scikit-learn provides a convenient classification report containing:

- Precision
- Recall
- F1-score
- Support


````python
from sklearn.metrics import classification_report

print(classification_report(
    y_test,
    y_pred,
    target_names=data.target_names
))
````

## 18. ROC-AUC

ROC-AUC is another important metric for binary classification.

The ROC curve examines the trade-off between:

- True Positive Rate
- False Positive Rate

AUC means **Area Under the Curve**.

General interpretation:

```text
AUC ≈ 1.0 → Excellent discrimination
AUC ≈ 0.5 → Similar to random guessing
```


````python
from sklearn.metrics import roc_curve, roc_auc_score

auc = roc_auc_score(y_test, y_prob)

fpr, tpr, thresholds = roc_curve(y_test, y_prob)

print(f"ROC-AUC: {auc:.4f}")

plt.figure(figsize=(8, 5))
plt.plot(fpr, tpr, label=f"ROC curve (AUC = {auc:.3f})")
plt.plot([0, 1], [0, 1], linestyle="--")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve")
plt.legend()
plt.grid(True)
plt.show()
````

## 19. Understanding the Decision Boundary

Logistic Regression converts the linear score `z` into a probability.

With a threshold of 0.5:

```text
Probability < 0.5 → Class 0
Probability ≥ 0.5 → Class 1
```

Since:

$$\sigma(z)=0.5$$

when:

$$z=0$$

the decision boundary occurs where:

$$b_0 + b_1x_1 + b_2x_2 + \cdots + b_nx_n = 0$$

This means Logistic Regression creates a **linear decision boundary** in the feature space.


## 20. Understanding Odds and Log-Odds

Probability and odds are related.

$$Odds = \frac{p}{1-p}$$

Logistic Regression models the **log-odds**:

$$log\left(\frac{p}{1-p}\right)
= b_0 + b_1x_1 + \cdots + b_nx_n$$

This is called the **logit**.

This explains why Logistic Regression is called a regression model even though it is used for classification: it models the log-odds as a linear function of the features.


## 21. Interpreting Coefficients with Odds Ratios

For a coefficient `β`:

$$Odds\ Ratio = e^{\beta}$$

For a one-unit increase in a feature, holding other features constant:

- `e^β > 1` → odds increase
- `e^β < 1` → odds decrease
- `e^β = 1` → no change in odds

Because we scaled the features in this example, interpret the coefficients as changes associated with approximately **one standard deviation** increase in the corresponding feature.


````python
coefficients = final_model.coef_[0]
odds_ratios = np.exp(coefficients)

feature_names = data.feature_names

top_features = sorted(
    zip(feature_names, coefficients, odds_ratios),
    key=lambda x: abs(x[1]),
    reverse=True
)

print("Top 10 features by absolute coefficient:\n")

for name, coef, odds_ratio in top_features[:10]:
    print(
        f"{name:30s} "
        f"coefficient={coef:8.4f} "
        f"odds_ratio={odds_ratio:8.4f}"
    )
````

## 22. Changing the Classification Threshold

The default threshold is commonly 0.5, but it does **not** have to be 0.5.

For example:

```python
threshold = 0.7
```

Then:

```text
Probability >= 0.7 → Class 1
Probability <  0.7 → Class 0
```

Increasing the threshold generally makes the model more conservative about predicting class 1.

The best threshold depends on the business or real-world cost of false positives and false negatives.


````python
threshold = 0.7

y_pred_07 = (y_prob >= threshold).astype(int)

print("Threshold:", threshold)
print("Accuracy:", accuracy_score(y_test, y_pred_07))
print("Precision:", precision_score(y_test, y_pred_07))
print("Recall:", recall_score(y_test, y_pred_07))
print("F1-score:", f1_score(y_test, y_pred_07))
````

# Summary :

### Logistic Regression

- It is primarily a **classification algorithm**.
- It is especially common for **binary classification**.
- It first calculates a linear score.
- The **Sigmoid function** converts the score into a probability.
- A threshold converts probability into a class.
- `predict()` returns class labels.
- `predict_proba()` returns class probabilities.
- Feature scaling can be useful.
- Confusion matrix helps understand different types of errors.
- Accuracy alone may not be enough for imbalanced problems.
- Precision, recall and F1-score provide additional insight.
- ROC-AUC measures ranking/discrimination performance.
- Logistic Regression produces a **linear decision boundary**.
- Coefficients describe changes in **log-odds**.
- `exp(coefficient)` gives an **odds ratio**.
