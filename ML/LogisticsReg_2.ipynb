```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn import metrics
```

```python
df = pd.read_csv("D:/Ashwath/NS_Trainings/MyStuff/gmatdata.csv")
print(df.head())
```

```python
X = df[['gmat', 'gpa','exp']]
y = df['admitted']
print(X)
print(y)
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.25,random_state=0,shuffle=True)
```

```python
print(X_train.shape)
print(y_train.shape)
print(X_test.shape)
print(y_test.shape)
```

```python
model = LogisticRegression()
model.fit(X_train,y_train)
y_pred=model.predict(X_train)
```

```python
print(y_train)
print(y_pred)
```

```python
print('Accuracy: ',(metrics.accuracy_score(y_train, y_pred))*100)
```

```python
#testing data
y_tpred=model.predict(X_test)
print(y_test)
print(y_tpred)
print('Accuracy: ',(metrics.accuracy_score(y_test, y_tpred))*100)
```
