```python
#knn implementation using sklearn
import pandas as pd
import numpy as np
```

```python
df = pd.DataFrame()
df = pd.read_csv("D:/Ashwath/NS_Trainings/MyStuff/naivebayesdata.txt")
print(df)
```

```python
from sklearn import preprocessing
#creating labelEncoder
le = preprocessing.LabelEncoder()
# Converting string labels into numbers.

weather_encoded=le.fit_transform(df['weather'])
print ("Weather:", weather_encoded)
```

```python
# Converting temp and labels into numbers
temp_encoded=le.fit_transform(df['temp'])
label=le.fit_transform(df['Play'])
```

```python
print ("Temp:",temp_encoded) #hot : 1, mild : 2, cool : 0
print ("Play:",label) # no : 0, yes : 1
```

```python
#combinig weather and temp into single listof tuples
features=list(zip(weather_encoded,temp_encoded))
print(features)
```

```python
from sklearn.neighbors import KNeighborsClassifier
model = KNeighborsClassifier(n_neighbors=5)
# Train the model using the training sets
model.fit(features,label)
```

```python
#Predict Output
predicted= model.predict([[0,2]]) # 0:Overcast, 2:Mild
print(predicted)
```

```python
#knn with user defined functions :

features1=list(zip(features,label))
print(features1)
```

```python
#define predict function

def predict(data,new): 
    dist =[]
    for v in data:
        attr = v[0]
        d = ((new - attr)**2).sum()**0.5
        #sqrt[(x1 - x2)^2 + (y1 - y2)^2]  
        dist.append((d,v[1]))
    dist.sort()
    k = 5   # taking 5 as nearst nieghbours
    
    x = dist[:k]       

    freq = {}
    for d in x:
        if freq.get(d[1])==None:
            freq[d[1]]=1
        else:
            freq[d[1]]+=1
    return (freq)
```

```python
newdata = np.array([0,2])
prevals = predict(features1,newdata)
print (prevals)
```
