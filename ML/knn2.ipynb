```python
#In the model the building part, we are using wine dataset, 
#which is a very famous multi-class classification problem. 
#This data is the result of a chemical analysis of wines grown 
#in the same region in Italy using three different cultivars. 
#The analysis determined the quantities of 13 constituents found 
#in each of the three types of wines.

#The dataset comprises 13 features ('alcohol', 'malic_acid', 
#'ash', 'alcalinity_of_ash', 
#'magnesium', 'total_phenols', 'flavanoids', 'nonflavanoid_phenols', '
#proanthocyanins', 
#'color_intensity', 'hue', 'od280/od315_of_diluted_wines', 'proline') 
#and a target (type of cultivars).

#This data has three types of cultivar classes: 'class_0', 
#'class_1', and 'class_2'. 
#Here, we need to  build a model to classify the type of cultivar. 
#The dataset is available in the scikit-learn library.
```

```python
#Import scikit-learn dataset library
from sklearn import datasets

#Load dataset
wine = datasets.load_wine()
```

```python
# print the names of the features
print(wine.feature_names)
```

```python
# print the label species(class_0, class_1, class_2)
print(wine.target_names)
```

```python
# print the wine data (top 5 records)
print(wine.data[0:5])
```

```python
# print the wine labels (0:Class_0, 1:Class_1, 2:Class_3)
print(wine.target)
```

```python
# print data(feature)shape
print(wine.data.shape)
```

```python
# Import train_test_split function
from sklearn.model_selection import train_test_split

# Split dataset into training set and test set
X_train, X_test, y_train, y_test = train_test_split(wine.data, wine.target, test_size=0.3) 
# 70% training and 30% test
```

```python
print(y_train)
print(y_test)
```

```python
#Import knearest neighbors Classifier model
from sklearn.neighbors import KNeighborsClassifier

#Create KNN Classifier
knn = KNeighborsClassifier(n_neighbors=9)

#Train the model using the training sets
knn.fit(X_train, y_train)
```

```python
#prediction and accuracy check for training data

ytpred = knn.predict(X_train)
print(ytpred)
```

```python
from sklearn import metrics
print("Accuracy:",metrics.accuracy_score(y_train, ytpred)*100)
```

```python
#Predict the response for test dataset
y_pred = knn.predict(X_test)
print(y_pred)
```

```python
#check accuracy
#Import scikit-learn metrics module for accuracy calculation
from sklearn import metrics
print("Accuracy:",metrics.accuracy_score(y_test, y_pred)*100)
```

```python
#k nearest centorid

from sklearn.neighbors import NearestCentroid

clf = NearestCentroid()

clf.fit(X_train, y_train)
```

```python
#prediction and accuracy check for training data using KNC

ytpred = clf.predict(X_train)
print(ytpred)
```

```python
from sklearn import metrics
print("Accuracy:",metrics.accuracy_score(y_train, ytpred)*100)
```
