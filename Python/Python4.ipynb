```python
# adding two list objects
# way 1:

a=[1,2,3,4,5]
b=[25,35,45,55,65]
i=0
res=[]  #initilizing of list : empty list 
for v in a:
    c=v+b[i] #1+25 = 26  2+35=37   3+45 = 48
    i+=1 #1
    res.append(c) #appends the data to the list object  [26,37,48,59,70]
print(res)
```

```python
# way 2:

a=[1,2,3,4,5]
b=[25,35,45,55,65]
res=[]

print(list(zip(a,b))) #creating list of tuples

for x,y in list(zip(a,b)): #[(1, 25), (2, 35), (3, 45), (4, 55), (5, 65)]
    res.append(x+y)
print(res)
```

```python
#zip(): 
#The zip function pairs the first elements of each iterator together, 
#then pairs the second elements together and so on.

#If the iterables in the zip function are not the same length, 
#then the smallest length iterable decides the length of the 
#generated output

a=('ram','shiva','krishna')
b=('sita','parvathi','radha')
x=list(zip(a,b))
print(x)
```

```python
#List Comprehension

#List comprehension offers a shorter syntax, 
#List comprehensions are used for creating new lists from 
#other iterables like tuples, strings, arrays, lists, etc. 
#A list comprehension consists of brackets containing the expression, 
#which is executed for each element along with the for loop 
#to iterate over each element.
```

```python
#list Comprehension method
kms = [200,300,350,600,750]
miles = [(float(x/1.6)) for x in kms]
print(miles)

#normal method
kms = [200,300,350,600,750]
m = []
for v in kms:
    m.append(float(v/1.6))
print(m)
```

```python
#finding the total salary for male employees and 
#female employees (aggregation)
name=['a','b','c','d','e']               
sal=[40000,50000,60000,55000,45000]
dno=[11,12,13,14,15]
gender=['m','m','f','f','m']
info=[]

res={}
for n,s,d,g in list(zip(name,sal,dno,gender)):
    info.append((n,g,s,d)) 
    
for v in info:
    print(v)
    
for v in info:  #('c', 'f', 60000, 13)
    gen = v[1] #f
    sal = v[2] #60000
    if res.get(gen)==None: 
        res[gen]=sal #{m:135000,f:115000}
    else:
        res[gen]+=sal
print (res)
```

```python
#program to classify the names of male employees and female employees
name=['rama','amar','rita','rani','giri','veni','somu','smitha']
sal=[40000,50000,60000,55000,45000,25000,40000,60000]
dno=[11,12,13,14,15,11,12,13,14]
gender=['m','m','f','f','m','f','m','f']
info=[]
res={}
for n,s,d,g in list(zip(name,sal,dno,gender)):  
    info.append((n,g,s,d))
    
#for v in info:
#    print(v)
    
for v in info:  
    gen = v[1] 
    nam = v[0] 
    if res.get(gen)==None:  #{m:[rama,amar],f:[rita]}
        res[gen]=[nam]     
    else:
        res[gen].append(nam)
print (res)

for k in res:
    print (k,res[k])
```
