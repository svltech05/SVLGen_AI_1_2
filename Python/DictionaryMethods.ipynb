```python
#dictionary methods : works only on dictionary
mydict = {101:1000,102:3000,103:4000}
print(mydict)
```

```python
#dict.clear() : empties the dictionary
mydict.clear()
print(mydict)
```

```python
#dict.items() returns a list of tuples containing the key-value 
#pairs in dict. The first item in each tuple is the key, and the 
#second item is the key’s value:

mydict = {101:1000,102:3000,103:4000}
print(mydict)
x = mydict.items()
print(x)
for i in x:
    print(i)
```

```python
for x in mydict:
    print(x,mydict[x])
```

```python
#keys() : returs all the keys from the given dictionary
mydict = {101:1000,102:3000,103:4000}
print(mydict.keys())
```

```python
#values() : returns all the values from the given dictionary
mydict = {101:1000,102:3000,103:4000}
print(mydict.values())
```

```python
for x in list(zip(mydict.keys(),mydict.values())):
    print(x)
```

```python
#dict.pop() : if the key is persent in the dicitonary
#it will remove that kvp from dictionary or else it gives an error

mydict = {101:1000,102:3000,103:4000}
print(mydict)

x = mydict.pop(102)
print(x)
print(mydict)
```

```python
#dict.popitem() 
#removes the last key-value pair added from d and returns it as a tuple:

mydict = {101:1000,102:3000,103:4000}
mydict.popitem()
print(mydict)
```

```python
#dict.update()
#merges the two dictionaries

#If the key is not present in dict, then key-value pair 
#from dict1 is added to dict

#If the key is already present in dict, the corresponding value 
#in dict for that key is updated to the value from dict1.

mydict = {101:1000,102:3000,103:4000}
#mydict1 = {104:4000,105:6000}
mydict1 = {103:8000,105:6000}
mydict.update(mydict1)
print(mydict)
```
