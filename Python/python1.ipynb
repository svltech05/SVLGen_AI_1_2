```python
custdict = {101:('rama',50000),102:('amar',25000),103:('rani',50000)}
trandict = {101:(5000,'cr'),102:(2000,'dr'),103:(6000,'dr')}
curbaldict = {}

for x in custdict:
    val = custdict[x]
    trn = trandict[x]
    if trn[1] == 'cr':
        curbal = val[1] + trn[0]
    elif trn[1]=='dr':
        curbal = val[1] - trn[0]
    curbaldict[x] = ((val[0],curbal))
print(curbaldict)
```
