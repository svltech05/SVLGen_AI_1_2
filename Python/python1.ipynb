```python
#compltex data type
# lists

#list is a collection of data itmes stored under a common name, 
#sequentially, seperated by comma.

#The most important characteristic of list is that 
#it is a mutable structure, 
#i.e changes can be made by references the list. 

#list is declared or represented by []
#starting index is zero (0) by default

#while sclicing starting index included, ending 
#index excluded

#a[starting index : ending index]          

a = [1,2,3,4,5,6,7.5,8,9,10]     
print (a)
```

```python
print(a[0])
print(a[:3])
print(a[5:])
print(a[1:7])
```

```python
print (a[-1])
print(a[-2])
print(a[-5:-2])
```

```python
#to show list is mutable
print(a)
a[5] = 101
print(a)
```

```python
#Dictionary

#Collection of Key Value Pairs
#Key Value pairs (KVP) is a set to two linked data items, 
#key is unique identifier
#val is data which is either the data that is 
#identified or a pointer to the location of that data

#dictionary is defined with {}

#dictionaryname = {key:value}

qual = {"BE":2020}
print(qual)

qual1 = {"sslc" : 2012, "puc" : 2014, "BSC" : 2017, "BE" : 2021}
print(qual1)
```

```python
#dictionary and tuple combination

plist = {'p1' : ('tv',40000,1),
           'p2' : ('laptop',60000,2),
            'p3' : ('mobile',10000,5)}
                    
print(plist)
print(type(plist))
```

```python
#dictionary and list combination

glist = {2014 : ['cricket','kabadhi'],
         2015 : ['Cricket','hockey'],
         2016 : ['football','cricket'],
         2017 : ['football','snooker']}
print (glist)
```

```python
#dicitionary, list and tuple combination

sales = {('branch1','p1') : [100,200],
         ('branch2','p1') : [50,70],
         ('branch1','p2') : [10,20],
         ('branch2','p2') : [5,7]
         }
print(sales)
```
