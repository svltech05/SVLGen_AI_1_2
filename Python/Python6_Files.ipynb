```python
#reading data from a file, convert it into tuple and 
#append it to a list object)

fileobj=open("D:/Ashwath/NS_Trainings/MyStuff/samplefile6.txt")

header = fileobj.readline()
print(header) 

flist = [] 
lines = fileobj.readlines()      #a = "45,m"

for line in lines:
    w=line.strip().split(',')  #w = [45,m]
    t = (int(w[0]),w[1]) #(45,m)
    flist.append(t)
#print(flist)
for v in flist:
    print(v)
```

```python
#reading data from a file, creating dicitionary by gender key)
fileobj=open("D:/Ashwath/NS_Trainings/MyStuff/samplefile6.txt")
header = fileobj.readline()
fdict = {}
lines = fileobj.readlines()
for line in lines:
    w=line.strip().split(',')
    gender = w[1]
    age = w[0]          
    if fdict.get(gender)==None: #{m:[45,34],f:[65]}
        fdict[gender] = [age]
    else:
        fdict[gender].append(age)
print(fdict)
```
