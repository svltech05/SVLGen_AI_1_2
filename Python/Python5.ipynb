```python
#function to transform department number to deptartment name

def fn_dno(x):
    if x == 11:
        dept = 'Administration'
    elif x == 12:
        dept = 'HR'
    elif x == 13:
        dept = 'Technical'
    elif x == 14:
        dept = 'Developmet'
    elif x == 15:
        dept = 'Testing'
    else:
        dept = 'Others'
    return dept
```

```python
#function to find the biggest element in the list object 
def fn_big(ylist):
    big = 0
    for i in ylist:
        if i > big:
            big = i
    return big
```

```python
a = [10,20,100,200,1000,6,4,2,1]
print(max(a))
print(fn_big(a))
```

```python
#function find the discount of an item:
def discount(itemprice):
    if itemprice >= 20000:
        damt = itemprice * (20/100)
        np = itemprice - damt
    elif itemprice >= 10000:
        damt = itemprice * (10/100)
        np = itemprice - damt    
    elif itemprice < 10000:
        damt = itemprice * (5/100)
        np=itemprice - damt
    return np
```

```python
ip = float(input("Enter item price : "))
netprice = discount(ip)
print(netprice)
```

```python
#function to convert the given string into proper case
def fupper(s):
    s=s.strip()     #s = " comPutEr "
    fc=s[0].upper()  #fc = "C"
    rc=s[1:].lower() #rc = "omputer"
    newstring=fc+rc  
    #newstring = Computer #adding/joining of two string variables 
    #concatenation
    return newstring
```

```python
#using fupper function:
nam =" comPuteR "
print (nam)
print(len(nam))
mname = fupper(nam)
print(mname)
print(len(mname))
```

```python
#fgender function to transform m : male f : female
def fgender(gender):
    if gender == 'm' or gender == 'M':
        res = "Male"
    if gender == 'f' or gender == 'F':
        res = "Female"
    return res
```

```python
name=['amar','barath','chitra','deeksha','eshwar']
sal=[40000,50000,60000,55000,45000]
dno=[11,12,13,14,15]
gender=['m','m','f','f','m']
info=[]
for n,s,d,g in list(zip(name,sal,dno,gender)):
    mname = fupper(n)
    gen = fgender(g)
    info.append((mname,gen,d,s))
#print(info)
for v in info:
    print(v)
```
