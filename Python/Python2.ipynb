```python
#if with logical operator
age = int(input("Enter your age "))
if age >= 60:       
    print("Senior citizen")
if age>40 and age<60: 
    print("Middle age")
if age>=20 and age<=40: 
    print("Young age")
if age>=13 and age < 20:
    print("Teen age")
if age < 13:
    print("Child")
```
