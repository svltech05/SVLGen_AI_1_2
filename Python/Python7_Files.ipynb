```python
def fupper(s):  #similiar to captalize(), converts first chracter to upper and rest to lower case
    s=s.strip()     
    fc=s[0].upper()  
    rc=s[1:].lower() 
    newstring=fc+rc  
    return newstring
```

```python
#Transforming name and gender ( m as Male and f as female)
fileobj=open("D:/Ashwath/NS_Trainings/MyStuff/samplefile3.txt")
line=fileobj.readline()
lines=fileobj.readlines()
for dline in lines:
    w=dline.strip().split(',')  #w = [101,amar,50000,m,11]
    uppername=fupper(w[1])
    if((w[-2])=='m'):
        gen="Male"
    if((w[-2])=='f'):
        gen="Female"
    w[-2]=gen
    w[1]=uppername
    print(w)
```

```python
#function to convert dno into dname
def fdname(dno):
    if dno == 11:
        dname = "Marketing"
    elif dno == 12:
        dname = "Finance"
    elif dno == 13:
        dname = "HR"
    elif dno == 14:
        dname = "Technical"
    else:
        dname = "Others"
    return dname
```

```python
#Single grouping multiple aggregation 

#(grouping by dno and finding sum,count,average,maximum and minimum)
fileobj=open("D:/Ashwath/NS_Trainings/MyStuff/samplefile3.txt")
head=fileobj.readline()
lines=fileobj.readlines()
res3={}
for line in lines:
    w=line.strip().split(',')
    mykey=w[-1] # grp by dno
    salary=int(w[2]) #11
    if res3.get(mykey)==None:
        res3[mykey]=[salary]  #{11:[50000,45000],12:[60000]}
    else:               
        res3[mykey].append(salary)   
#print (res3)

for x in res3:
    print(x,res3[x])
```

```python
finres={}
for x in list(zip(res3.keys(),res3.values())):
    print(x)
```

```python
finres={}
for k,v in list(zip(res3.keys(),res3.values())): 
    #[(11,(50000,60000....),(12,(.....)))]
    tot=sum(v)
    cnt=len(v)
    avg=tot/cnt
    mx=max(v)
    mn=min(v)
    r=(tot,cnt,avg,mx,mn)
    finres[k]=r 
#print(finres)
for x in finres:
    print(x,finres[x])
```

```python
#multiple grouping multiple aggregation 
#(grouping by dno,gender and finding sum,count,average,maximum and minimum)

fileobj=open("D:/Ashwath/NS_Trainings/MyStuff/samplefile3.txt")
head=fileobj.readline()
lines=fileobj.readlines()
res4={}
for line in lines:
    w=line.strip().split(',')
    mykey=(w[-1],w[-2])
    salary=int(w[2])
    if res4.get(mykey)==None: 
        res4[mykey]=[salary]  #{(11,m) : [45000,50000,50000]}
    else:
        res4[mykey].append(salary)
#print(res4)

finres={}
for k,v in list(zip(res4.keys(),res4.values())):
    tot=sum(v)
    cnt=len(v)
    avg=tot/cnt
    mx=max(v)
    mn=min(v)
    r=(tot,cnt,avg,mx,mn)
    finres[k]=r
#print(finres)
for k,v in list(zip(finres.keys(),finres.values())):
    print(k,v)
```
